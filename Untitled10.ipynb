```python
# ==========================
# IMPROVED AI CHATBOT TRAINING
# ==========================

import random
import numpy as np
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import LabelEncoder

# Expanded dataset (VERY IMPORTANT)
intents = {
    "intents": [
        {
            "tag": "greeting",
            "patterns": [
                "Hi", "Hello", "Hey", "Hey there",
                "Good morning", "Good evening",
                "hello!", "hi!", "hello bot"
            ],
            "responses": [
                "Hello! How can I help you today?",
                "Hi there! What can I do for you?"
            ]
        },
        {
            "tag": "goodbye",
            "patterns": [
                "Bye", "See you", "Goodbye",
                "Exit", "bye bye", "see you later"
            ],
            "responses": [
                "Goodbye! Have a great day!",
                "See you soon!"
            ]
        },
        {
            "tag": "joke",
            "patterns": [
                "Tell me a joke",
                "Make me laugh",
                "Say something funny",
                "joke please",
                "funny joke"
            ],
            "responses": [
                "Why don't skeletons fight each other? They don't have the guts!"
            ]
        },
        {
            "tag": "machine_learning",
            "patterns": [
                "What is machine learning?",
                "Explain ML",
                "Define machine learning",
                "What is AI?",
                "Explain artificial intelligence"
            ],
            "responses": [
                "Machine Learning is a field of AI where systems learn from data and improve automatically."
            ]
        }
    ]
}

# Prepare training data
sentences = []
labels = []
responses = {}

for intent in intents["intents"]:
    for pattern in intent["patterns"]:
        sentences.append(pattern.lower())
        labels.append(intent["tag"])
    responses[intent["tag"]] = intent["responses"]

# Encode labels
label_encoder = LabelEncoder()
encoded_labels = label_encoder.fit_transform(labels)

# 🔥 IMPORTANT CHANGE: Add ngram_range
vectorizer = TfidfVectorizer(ngram_range=(1,2))
X = vectorizer.fit_transform(sentences)

# Train model
model = LogisticRegression(max_iter=500)
model.fit(X, encoded_labels)

print("✅ Improved Model Trained Successfully!")
```

```python
# ==========================
# RUN IMPROVED CHATBOT
# ==========================

print("\n🤖 AI Chatbot is Ready! Type 'exit' to stop.\n")

while True:
    user_input = input("You: ")

    if user_input.lower() == "exit":
        print("Bot: Goodbye! 👋")
        break

    X_input = vectorizer.transform([user_input.lower()])
    prediction = model.predict(X_input)
    probabilities = model.predict_proba(X_input)

    confidence = max(probabilities[0])
    tag = label_encoder.inverse_transform(prediction)[0]

    # 🔥 Lowered confidence threshold
    if confidence > 0.30:
        reply = random.choice(responses[tag])
    else:
        reply = "I'm not sure I understood that. Can you rephrase?"

    print(f"Bot: {reply}")
    print(f"(Confidence: {confidence:.2f})\n")
```
